```python
Path= "C:/Users/BeaBa/Documents/GitHub/si/datasets/iris/"
iris = read_csv(Path + "iris.csv", sep=",", label=True)
iris.summary()
```

```python
iris_pca = PCA(n_components=2)
iris_pca.fit(iris)
```

```python
# transformation of the dataset to the new principal components
iris_transformed = iris_pca.transform(iris)
iris_transformed.shape()
```
